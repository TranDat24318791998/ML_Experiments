```python
class Circle():
    def __init__(self, radius, id):
        self.radius = 10
        self._id = id

    @property
    def area(self):
        return 3.14*self.radius**2

    def perimeter(self):
        return 2*3.14*self.radius

    @property
    def get_id(self):
        return self._id

circle = Circle(10, 123)
circle.area, circle.perimeter()
```

```python
from typing import Protocol

class Named(Protocol):
    name: str                 # attribute đọc/ghi

class HasId(Protocol):
    @property
    def id(self) -> int: ...  # chỉ yêu cầu đọc được (read-only)

class User:
    def __init__(self, name: str, id: int):
        self.name = name
        self._id = id

    @property
    def id(self) -> int:
        return self._id

def greet(x: Named) -> str:
    return f"Hi {x.name}"

greet(User("Dat", 1))  # OK
```
