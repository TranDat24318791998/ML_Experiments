```python
class Circle():
    def __init__(self, radius, id):
        self.radius = 10
        self._id = id

    @property
    def area(self):
        return 3.14*self.radius**2

    def perimeter(self):
        return 2*3.14*self.radius

    @property
    def get_id(self):
        return self._id

circle = Circle(10, 123)
circle.area, circle.perimeter()
```

```python
from typing import Protocol

class Named(Protocol):
    name: str                 # attribute đọc/ghi

class HasId(Protocol):
    @property
    def id(self) -> int: ...  # chỉ yêu cầu đọc được (read-only)

class User:
    def __init__(self, name: str, id: int):
        self.name = name
        self._id = id

    @property
    def id(self) -> int:
        return self._id

def greet(x: Named) -> str:
    return f"Hi {x.name}"

greet(User("Dat", 1))  # OK
```

```python
from typing import Literal
from pydantic import ConfigDict, Field

class Profile(BaseModel):
    model_config = ConfigDict(strict=True)
    score: int
    sport: Literal["football", "tenis", "badminton"]

class User(BaseModel):
    name: str
    id: int = Field(frozen=True)
    addr: str
    age: int = Field(ge=20, le=60)
    profile: Profile
        

user = User(name="Dat", id="123", addr="HCM", age=18, profile=Profile(score=10, sport="football", height=170))
user.name, user.id, user.addr, user.age, user.profile
```

```python
class Profile():
    score: float
    sports: str #[football, tenis, badminton] - volleyball: reject

class User():
    def __init__(
            self, 
            name:str, 
            id:int, 
            addr:str, 
            age:str, 
            profile:Profile) -> None:
        self.name = name
        self.id = int(id)
        self.addr = addr
        self.age = age
        # self.profile = profile if isinstance(profile, Profile) else None
        if isinstance(profile, Profile):
            self.profile = profile
        else:
            raise ValueError("Invalid profile")
        

user = User(name="Dat", id="123", addr="HCM", age=28, profile="profile")
user.name, user.id, user.addr, user.age, user.profile
```
