```python
from enum import Enum, StrEnum
```

```python
# order status: paid, pending, cancelled
# Without enum
status = 'paid'

if status == 'paid':
    print("Done")
elif status == "pending":
    print("Pending!!")
```

```python
class OrderStatus():
    PAID = 'paid'
    PENDING = 'pending'
    CANCELLED = 'cancelled'

status = 'paid'
if status not in (
    OrderStatus.PAID,
    OrderStatus.PENDING,
    OrderStatus.CANCELLED
):
    raise ValueError("Status is not valid")
```

```python
class OrderStatus(Enum):
    PAID = 'paid'
    PENDING = 'pending'
    CANCELLED = 'cancelled'

type(OrderStatus.PAID)
status = OrderStatus('paid')
# status == OrderStatus.PAID
# status == 'paid'
OrderStatus.PAID == 'paid'
print()
```

```python
class OrderStatus(StrEnum):
    PAID = 'paid'
    PENDING = 'pending'
    CANCELLED = 'cancelled'

type(OrderStatus.PAID)
status = OrderStatus('paid')
# status == OrderStatus.PAID
# status == 'paid'
OrderStatus.PAID == 'paid'
```

```python
try:
    status = OrderStatus('paid')
except:
    raise ValueError('error')

print(status, type(status))
print(status == OrderStatus.PAID)
print(status == 'paid')
```
