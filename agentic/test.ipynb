```python
import langchain
import langgraph

from langchain.chat_models import init_chat_model
```

```python
llm = init_chat_model(
    model="qwen2.5:7b",
    model_provider="ollama",
    base_url="http://localhost:11434", 
    temperature=0,
)
```

```python
from langchain.agents import create_agent
from pydantic import BaseModel

def search() -> str:
    """ Search capital of Vietnam """
    return "Ho Chi Minh"

def say_hi(name:str) -> str:
    """Say hi with a person"""
    return f"Hi {name}, how are you doing to day, everything is oke?"

def div(a:float, b:float) -> float:
    """Division operation"""
    return a / b

agent = create_agent(
    model=llm,
    tools=[search, say_hi, div],
    system_prompt="You are a helpful assistant"
)

result = agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "Calculate 12345/8732, exactly 10 decimal"
            }
        ]
    }
)

result['messages'][-1].content
```
